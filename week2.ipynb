```python
# 1. Drop rows with missing datetime (Crucial: 1,551 rows removed)
ldf = ldf.dropna(subset=['datetime']).reset_index(drop=True)
```

```python
# 2. Convert datetime formats safely
ldf['datetime'] = pd.to_datetime(ldf['datetime'])
ldf['sunrise'] = pd.to_datetime(ldf['sunrise'], errors='coerce')
ldf['sunset'] = pd.to_datetime(ldf['sunset'], errors='coerce')
```

```python
# 4. Fill rain/snow categorical and numeric gaps
ldf['preciptype'] = ldf['preciptype'].fillna('none')
ldf['snow'] = ldf['snow'].fillna(0.0)
ldf['snowdepth'] = ldf['snowdepth'].fillna(0.0)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score
```

```python

# 1. Classification Target: Will it rain tomorrow? (Shift binary status back 1 step)
ldf['will_rain_tomorrow'] = ldf['is_rain'].shift(-1)

# 2. Time-Series Target: What will the temperature be tomorrow? (Shift continuous temp back 1 step)
ldf['next_day_temp'] = ldf['temp'].shift(-1)

# 3. Clean Boundary Rows: Drop the final row since it has no "tomorrow" to calculate against
ldf = ldf.dropna(subset=['will_rain_tomorrow', 'next_day_temp'])
f = ldf.dropna()
```

```python
from sklearn.model_selection import train_test_split

# --- SPLIT A: Classification Matrix ---
# Drop both target columns to protect the input features (X)
X_class = ldf.drop(columns=['will_rain_tomorrow', 'next_day_temp'])
y_class = ldf['will_rain_tomorrow'].astype(int)

# Standard random split (fine for evaluating independent classification events)
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class, y_class, test_size=0.2, random_state=42
)

# --- SPLIT B: Time-Series Matrix ---
X_ts = ldf.drop(columns=['will_rain_tomorrow', 'next_day_temp'])
y_ts = ldf['next_day_temp']

# Chronological split index (Strictly preserves time sequence; no shuffling allowed)
split_idx = int(len(ldf) * 0.8)

X_train_ts = X_ts.iloc[:split_idx]
y_train_ts = y_ts.iloc[:split_idx]
X_test_ts = X_ts.iloc[split_idx:]
y_test_ts = y_ts.iloc[split_idx:]
```

```python
# 1. Convert Sunrise and Sunset into pure numeric hours (e.g., 6.25 instead of 06:15 AM)
# This turns timestamps into numbers your model can actually learn from
if 'sunrise' in ldf.columns and ldf['sunrise'].dtype == 'datetime64[ns]':
    ldf['sunrise_hour'] = ldf['sunrise'].dt.hour + (ldf['sunrise'].dt.minute / 60.0)
    ldf = ldf.drop(columns=['sunrise'])

if 'sunset' in ldf.columns and ldf['sunset'].dtype == 'datetime64[ns]':
    ldf['sunset_hour'] = ldf['sunset'].dt.hour + (ldf['sunset'].dt.minute / 60.0)
    ldf = ldf.drop(columns=['sunset'])

# 2. Re-create your feature matrix (X) by dropping ALL remaining datetime columns
# Note: Since 'datetime' is your dataframe index, it won't be harmed here
X_class = ldf.drop(columns=['will_rain_tomorrow', 'next_day_temp'])
y_class = ldf['will_rain_tomorrow'].astype(int)

# 3. Explicitly filter out any stray datetime columns just to be 100% safe
X_class = X_class.select_dtypes(exclude=['datetime64', 'datetime64[ns]', 'object'])

# 4. Re-split your classification data with the clean numeric features
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_class, y_class, test_size=0.2, random_state=42
)

# 5. Re-split your time-series data with the clean numeric features
X_ts = ldf.drop(columns=['will_rain_tomorrow', 'next_day_temp']).select_dtypes(exclude=['datetime64', 'datetime64[ns]', 'object'])
y_ts = ldf['next_day_temp']

X_train_ts = X_ts.iloc[:split_idx]
y_train_ts = y_ts.iloc[:split_idx]
X_test_ts = X_ts.iloc[split_idx:]
y_test_ts = y_ts.iloc[split_idx:]

print("Stray datetime columns removed. Features are now purely numeric!")
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# 1. Random Forest Classifier
rf_class_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_class_model.fit(X_train_c, y_train_c)
rf_class_preds = rf_class_model.predict(X_test_c)

print("=== RANDOM FOREST CLASSIFICATION REPORT ===")
print(classification_report(y_test_c, rf_class_preds))

# 2. Linear Classifier (Logistic Regression)
# Note: Weather variables have different ranges; we increase max_iter so the linear optimizer converges safely
lr_class_model = LogisticRegression(max_iter=5000, random_state=42)
lr_class_model.fit(X_train_c, y_train_c)
lr_class_preds = lr_class_model.predict(X_test_c)

print("\n=== LINEAR CLASSIFICATION (LOGISTIC REGRESSION) REPORT ===")
print(classification_report(y_test_c, lr_class_preds))
```
